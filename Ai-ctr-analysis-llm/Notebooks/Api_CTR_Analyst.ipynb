## 1. OpenAI Completions Usage API Extended 
----
源代码： https://github.com/openai/openai-cookbook/blob/main/examples/completions_usage_api.ipynb

数据源： https://tianchi.aliyun.com/dataset/56 


```python
from openai import OpenAI

# 初始化客户端
client = OpenAI()

# 存储对话历史（非常关键）
messages = [
    {"role": "system", 
     "content": "你是一个数据分析助手，回答要专业、简洁"}
]

print("🤖 AI Chat 已启动（输入 exit 退出）")

while True:
    user_input = input("\n你：")

    if user_input.lower() == "exit":
        break

    # 添加用户输入
    messages.append({"role": "user", "content": user_input})

    # 调用 OpenAI API
    response = client.chat.completions.create(
        model="gpt-4.1-mini",
        messages=messages
    )

    reply = response.choices[0].message.content

    print("\nAI：", reply)

    # 把 AI 回复加入历史（关键！）
    messages.append({"role": "assistant", "content": reply})
```
